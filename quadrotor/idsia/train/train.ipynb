# IDSIA models

Three networks, one recipe. Train/val/test is Square–Random–Chirp runs 1–3 / run 4 / Melon. Windows are 50 steps at 100 Hz, stride 1, WeightedMSE $\lambda=0.1$, Adam $10^{-5}\to10^{-8}$ cosine $T=500$, 20 epochs.

Build off their RK4 leftover first (`PhysQuotient`, then `PhysQuotientSym`). `SymmetryQuadDynamicsNet` is the symmetry-only test.

Checkpoints: `out/runs/<family>/seed_k/{best.pt,last.pt}`. Training resumes `last.pt` if it exists.

```python
SEED = 0  # change to start or resume another copy, for eval we use seeds 0, 1, 2, 3, 4 for all models
DEVICE = "mps"  # train_device() will upgrade to mps when DEVICE is cpu and mps exists
```

## PhysQuotient

Frozen Crazyflie RK4. Residual reads the 13-D quotient $(g^{\mathrm{body}}, v^{\mathrm{body}}, \omega, \Omega)$ and writes body $(\Delta p,\Delta v,\Delta\theta,\Delta\omega)$. No orbit.

```python
from quadrotor.geometry import train_device
from quadrotor.idsia.train.lib import train_family
from quadrotor.idsia.utils.idsia_training_reproduction_params import PHYS_QUOTIENT

train_family(PHYS_QUOTIENT, SEED, device=train_device(DEVICE))
```

## PhysQuotientSym

Same frozen RK4 and body residual as PhysQuotient, Reynolds-averaged over the four Crazyflie-X mirrors.

```python
from quadrotor.idsia.utils.idsia_training_reproduction_params import PHYS_QUOTIENT_SYM

train_family(PHYS_QUOTIENT_SYM, SEED, device=train_device(DEVICE))
```

## SymmetryQuadDynamicsNet

No analytic Crazyflie model. Same quotient + Z₂×Z₂ leftover, integrated from the current state. Tests whether symmetry can stand in for their physics prior.

```python
from quadrotor.idsia.utils.idsia_training_reproduction_params import SYM_QUAD_DYNAMICS

train_family(SYM_QUAD_DYNAMICS, SEED, device=train_device(DEVICE))
```
